## 8: Missing Value Profile Analysis
Analyze missing value counts and percentages across all dataset columns, visualizing the result with an interactive Plotly bar chart.

```python

# Dataset Missing Value Analysis — Kaggle-Safe Rendering

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# Step 1: Validate the dataset
if "df" not in globals():
    print("ERROR: DataFrame 'df' was not found. Run the dataset-loading cell first.")

elif not isinstance(df, pd.DataFrame):
    print("ERROR: 'df' is not a pandas DataFrame.")

elif df.empty:
    print("The dataset is empty.")

else:
    # Step 2: Calculate missing values
    missing_df = pd.DataFrame({
        "Feature": df.columns,
        "Missing_Count": df.isnull().sum().values,
        "Missing_Pct": (df.isnull().mean().values * 100)
    })

    missing_df = (
        missing_df[missing_df["Missing_Count"] > 0]
        .sort_values("Missing_Pct", ascending=False)
        .reset_index(drop=True)
    )

    display(Markdown("## Dataset Missing Value Analysis"))

    # Step 3: Handle datasets without missing values
    if missing_df.empty:
        print("No missing values were found.")
        print(f"Total rows: {len(df):,}")
        print(f"Total columns: {len(df.columns):,}")

    else:
        # Step 4: Display the results table
        display(missing_df)

        # Step 5: Draw a static Matplotlib chart
        chart_height = max(4, min(12, len(missing_df) * 0.4))

        fig, ax = plt.subplots(figsize=(12, chart_height))

        bars = ax.barh(
            missing_df["Feature"].astype(str),
            missing_df["Missing_Pct"],
            edgecolor="black",
            linewidth=0.4
        )

        ax.set_title(
            "Missing Values by Feature",
            fontsize=15,
            fontweight="bold"
        )
        ax.set_xlabel("Missing Values (%)")
        ax.set_ylabel("Feature")
        ax.grid(axis="x", linestyle="--", alpha=0.3)
        ax.set_axisbelow(True)

        max_pct = missing_df["Missing_Pct"].max()
        ax.set_xlim(0, max(5, min(115, max_pct * 1.2)))

        for bar, pct in zip(bars, missing_df["Missing_Pct"]):
            ax.annotate(
                f"{pct:.2f}%",
                (bar.get_width(), bar.get_y() + bar.get_height() / 2),
                xytext=(4, 0),
                textcoords="offset points",
                va="center",
                fontsize=9
            )

        fig.tight_layout()
        plt.show()
        plt.close(fig)

    # Step 6: Print a summary
    total_missing = int(df.isnull().sum().sum())
    total_cells = df.size
    overall_pct = (
        total_missing / total_cells * 100
        if total_cells > 0 else 0
    )

    print("\nMissing Value Summary")
    print("-" * 40)
    print(f"Total rows: {len(df):,}")
    print(f"Total columns: {len(df.columns):,}")
    print(f"Features with missing values: {len(missing_df):,}")
    print(f"Total missing cells: {total_missing:,}")
    print(f"Overall missing-cell percentage: {overall_pct:.2f}%")
```

## 12: Age Distribution Histogram
Visualize the distribution of patient age using an interactive Plotly histogram with marginal box plot.

```python

# Patient Age Distribution — Kaggle-Safe Visualization

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# Step 1: Validate the dataset
if "df" not in globals() or not isinstance(df, pd.DataFrame):
    print("ERROR: DataFrame 'df' is unavailable. Run the dataset-loading cell first.")

elif df.empty:
    print("The dataset is empty.")

else:
    # Step 2: Identify the age column
    age_col = next(
        (col for col in df.columns if str(col).lower().strip() == "age"),
        None
    )

    # Fallback: Find a suitable numeric column if age is unavailable
    if age_col is None:
        numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

        age_col = (
            numeric_cols[1]
            if len(numeric_cols) > 1
            else numeric_cols[0]
            if numeric_cols
            else None
        )

    if age_col is None:
        print("No suitable numeric column was found for age distribution.")

    else:
        # Step 3: Clean the selected column
        age_data = pd.to_numeric(df[age_col], errors="coerce")
        age_data = age_data.replace([np.inf, -np.inf], np.nan).dropna()

        display(Markdown(f"## Patient Age Distribution ({age_col})"))

        if age_data.empty:
            print(f"No valid numeric observations found in '{age_col}'.")

        else:
            # Step 4: Display descriptive statistics
            print("Age Distribution Summary")
            print("-" * 40)
            print(f"Selected column: {age_col}")
            print(f"Valid observations: {len(age_data):,}")
            print(f"Missing/invalid observations: {len(df) - len(age_data):,}")
            print(f"Mean: {age_data.mean():.2f}")
            print(f"Median: {age_data.median():.2f}")
            print(f"Minimum: {age_data.min():.2f}")
            print(f"Maximum: {age_data.max():.2f}")
            print(f"Standard deviation: {age_data.std():.2f}")

            # Step 5: Generate static visualizations
            try:
                fig, axes = plt.subplots(
                    2,
                    1,
                    figsize=(12, 7),
                    gridspec_kw={"height_ratios": [3, 1]}
                )

                # Histogram
                axes[0].hist(
                    age_data,
                    bins="auto",
                    edgecolor="black",
                    alpha=0.85
                )

                axes[0].set_title(
                    f"Patient Age Distribution ({age_col})",
                    fontsize=15,
                    fontweight="bold"
                )
                axes[0].set_xlabel("Age")
                axes[0].set_ylabel("Number of Observations")
                axes[0].grid(axis="y", linestyle="--", alpha=0.3)

                # Box plot
                axes[1].boxplot(
                    age_data,
                    vert=False,
                    showmeans=True
                )

                axes[1].set_xlabel("Age")
                axes[1].set_yticks([])
                axes[1].set_title("Age Distribution Box Plot")
                axes[1].grid(axis="x", linestyle="--", alpha=0.3)

                fig.tight_layout()
                plt.show()
                plt.close(fig)

            except Exception as plot_error:
                print("Visualization failed:", str(plot_error))

            # Step 6: Generate an age frequency table
            try:
                print("\nAge Frequency Table")
                print("-" * 40)

                unique_values = age_data.nunique()

                if unique_values == 1:
                    frequency_table = pd.DataFrame({
                        "Age Range": [str(age_data.iloc[0])],
                        "Patient Count": [len(age_data)]
                    })
                else:
                    bins = min(10, unique_values)

                    age_bins = pd.cut(
                        age_data,
                        bins=bins,
                        duplicates="drop"
                    )

                    frequency_table = (
                        age_bins.value_counts(sort=False)
                        .rename_axis("Age Range")
                        .reset_index(name="Patient Count")
                    )

                display(frequency_table)

            except Exception as table_error:
                print("Frequency table could not be generated.")
                print("Error details:", str(table_error))
```

## 14: Age Spread by Diagnosis Result
Examine age variations across diagnostic categories using an interactive Plotly box plot.

```python

# Age Spread Across Categories — Kaggle-Safe Visualization

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# Step 1: Validate the dataset
if "df" not in globals() or not isinstance(df, pd.DataFrame):
    print("ERROR: DataFrame 'df' is unavailable. Run the dataset-loading cell first.")

elif df.empty:
    print("The dataset is empty.")

else:
    # Step 2: Identify the age column
    age_col = next(
        (c for c in df.columns if str(c).strip().lower() == "age"),
        None
    )

    # Step 3: Identify suitable categorical columns
    cat_candidates = [
        c for c in df.select_dtypes(
            include=["object", "category", "bool"]
        ).columns
        if df[c].nunique(dropna=True) < 20
    ]

    # Prefer the diagnosis_result column
    diag_col = next(
        (
            c for c in df.columns
            if str(c).strip().lower() == "diagnosis_result"
        ),
        None
    )

    # Use another categorical column if diagnosis_result is unavailable
    if diag_col is None:
        diag_col = cat_candidates[0] if cat_candidates else None

    if age_col is None or diag_col is None:
        print("A suitable age column or categorical grouping column was not found.")
        print("Available columns:", df.columns.tolist())

    else:
        # Step 4: Clean the data
        plot_df = df[[diag_col, age_col]].copy()

        plot_df[age_col] = pd.to_numeric(
            plot_df[age_col],
            errors="coerce"
        )

        plot_df[age_col] = plot_df[age_col].replace(
            [np.inf, -np.inf],
            np.nan
        )

        plot_df = plot_df.dropna(subset=[diag_col, age_col])

        display(
            Markdown(f"## Age Spread Across Categories ({diag_col})")
        )

        if plot_df.empty:
            print("No valid records are available for this comparison.")

        else:
            # Step 5: Order categories by sample size
            categories = (
                plot_df[diag_col]
                .value_counts()
                .index
                .tolist()
            )

            grouped_ages = [
                plot_df.loc[
                    plot_df[diag_col] == category,
                    age_col
                ].values
                for category in categories
            ]

            # Step 6: Calculate category-wise statistics
            summary_df = (
                plot_df.groupby(diag_col, observed=True)[age_col]
                .agg(
                    Patient_Count="count",
                    Mean_Age="mean",
                    Median_Age="median",
                    Minimum_Age="min",
                    Maximum_Age="max"
                )
                .round(2)
                .sort_values("Patient_Count", ascending=False)
                .reset_index()
            )

            print(f"Grouping column: {diag_col}")
            print(f"Age column: {age_col}")
            print(f"Valid records: {len(plot_df):,}")
            print(f"Number of categories: {len(categories)}")

            display(summary_df)

            # Step 7: Create a static box plot
            try:
                fig_width = max(
                    10,
                    min(16, len(categories) * 1.2)
                )

                fig, ax = plt.subplots(
                    figsize=(fig_width, 6)
                )

                ax.boxplot(
                    grouped_ages,
                    labels=[str(c) for c in categories],
                    patch_artist=True,
                    showmeans=True,
                    showfliers=True
                )

                ax.set_title(
                    f"Age Spread Across Categories ({diag_col})",
                    fontsize=14,
                    fontweight="bold",
                    pad=15
                )

                ax.set_xlabel(str(diag_col), fontsize=11)
                ax.set_ylabel(str(age_col), fontsize=11)
                ax.grid(axis="y", linestyle="--", alpha=0.3)
                ax.set_axisbelow(True)

                plt.setp(
                    ax.get_xticklabels(),
                    rotation=30,
                    ha="right"
                )

                fig.tight_layout()
                plt.show()
                plt.close(fig)

            except Exception as plot_error:
                print("Box plot rendering failed.")
                print("The summary table remains available.")
                print("Error details:", str(plot_error))

            # Step 8: Print a readable summary
            print("\nCategory-wise Age Summary")
            print("-" * 45)

            for _, row in summary_df.iterrows():
                print(
                    f"{row[diag_col]}: "
                    f"n={int(row['Patient_Count'])}, "
                    f"mean age={row['Mean_Age']:.2f}, "
                    f"median age={row['Median_Age']:.2f}"
                )
```

## 15: Correlation Matrix Heatmap
Compute pairwise correlation coefficients across numeric features and plot an interactive heatmap.

```python

# Numeric Feature Correlation Matrix — Kaggle-Safe Visualization

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# Step 1: Validate the dataset
if "df" not in globals() or not isinstance(df, pd.DataFrame):
    print("ERROR: DataFrame 'df' is unavailable. Run the dataset-loading cell first.")

elif df.empty:
    print("The dataset is empty.")

else:
    # Step 2: Select numeric columns
    numeric_df = df.select_dtypes(include=[np.number]).copy()

    # Remove constant columns
    numeric_df = numeric_df.loc[
        :, numeric_df.nunique(dropna=True) > 1
    ]

    display(Markdown("## Numeric Feature Correlation Matrix"))

    if numeric_df.shape[1] < 2:
        print("Insufficient numeric features for a correlation matrix.")
        print(f"Usable numeric features: {numeric_df.shape[1]}")

    else:
        # Step 3: Calculate the correlation matrix
        corr = numeric_df.corr(method="pearson")
        corr = corr.replace([np.inf, -np.inf], np.nan)

        print(f"Numeric features analyzed: {len(corr.columns)}")
        print(f"Dataset rows: {len(df):,}")

        # Display the numerical correlation matrix
        display(corr.round(3))

        # Step 4: Create a static heatmap
        try:
            n_features = len(corr.columns)
            fig_size = max(8, min(18, n_features * 0.65))

            fig, ax = plt.subplots(
                figsize=(fig_size, fig_size * 0.85)
            )

            corr_values = corr.to_numpy(dtype=float)
            masked_corr = np.ma.masked_invalid(corr_values)

            image = ax.imshow(
                masked_corr,
                cmap="coolwarm",
                vmin=-1,
                vmax=1,
                aspect="auto",
                interpolation="nearest"
            )

            labels = corr.columns.astype(str).tolist()

            ax.set_xticks(range(n_features))
            ax.set_yticks(range(n_features))

            ax.set_xticklabels(
                labels,
                rotation=90,
                fontsize=max(6, min(10, 150 // n_features))
            )

            ax.set_yticklabels(
                labels,
                fontsize=max(6, min(10, 150 // n_features))
            )

            ax.set_title(
                "Numeric Feature Correlation Matrix",
                fontsize=15,
                fontweight="bold",
                pad=18
            )

            # Annotate cells when the matrix is small enough
            if n_features <= 20:
                for i in range(n_features):
                    for j in range(n_features):
                        value = corr_values[i, j]

                        if np.isfinite(value):
                            text_color = (
                                "white" if abs(value) > 0.55 else "black"
                            )

                            ax.text(
                                j, i, f"{value:.2f}",
                                ha="center",
                                va="center",
                                fontsize=8,
                                color=text_color
                            )

            colorbar = fig.colorbar(
                image,
                ax=ax,
                fraction=0.046,
                pad=0.04
            )
            colorbar.set_label("Pearson Correlation")

            fig.tight_layout()
            plt.show()
            plt.close(fig)

        except Exception as plot_error:
            print("Heatmap rendering failed.")
            print("The numerical correlation matrix remains available.")
            print("Error details:", str(plot_error))

        # Step 5: Find the strongest feature relationships
        try:
            upper_triangle = corr.where(
                np.triu(
                    np.ones(corr.shape, dtype=bool),
                    k=1
                )
            )

            strongest_pairs = (
                upper_triangle.stack()
                .rename("Correlation")
                .reset_index()
            )

            strongest_pairs.columns = [
                "Feature 1",
                "Feature 2",
                "Correlation"
            ]

            strongest_pairs["Absolute Correlation"] = (
                strongest_pairs["Correlation"].abs()
            )

            strongest_pairs = (
                strongest_pairs.sort_values(
                    "Absolute Correlation",
                    ascending=False
                )
                .head(10)
                .drop(columns="Absolute Correlation")
                .reset_index(drop=True)
            )

            print("\nTop 10 Strongest Feature Relationships")

            if strongest_pairs.empty:
                print("No valid feature pairs were available.")
            else:
                display(strongest_pairs.round(3))

        except Exception as summary_error:
            print("Could not calculate the strongest feature relationships.")
            print("Error details:", str(summary_error))
```

## 16: Outlier Detection via IQR Screening
Identify statistical outliers across primary numeric features using the Interquartile Range (IQR) method.

```python
if not df.empty:
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    outlier_summary = {}
    for col in numeric_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
        outlier_summary[col] = len(outliers)
    
    outlier_df = pd.DataFrame(list(outlier_summary.items()), columns=['Feature', 'Outlier_Count'])
    display(outlier_df.sort_values(by='Outlier_Count', ascending=False).head(10))
else:
    print("Dataset empty.")
```

```python

# 17. Feature Aggregations & Group Means — Kaggle-Safe Visualization

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# Step 1: Validate the dataset
if "df" not in globals() or not isinstance(df, pd.DataFrame):
    print("ERROR: DataFrame 'df' is unavailable. Run the dataset-loading cell first.")

elif df.empty:
    print("The dataset is empty.")

else:
    display(Markdown("## Feature Aggregations & Group Means"))

    # Step 2: Identify the age column
    age_col = next(
        (c for c in df.columns if str(c).strip().lower() == "age"),
        None
    )

    # Step 3: Find suitable categorical columns
    categorical_cols = df.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()

    valid_cat_cols = [
        c for c in categorical_cols
        if 1 < df[c].nunique(dropna=True) < 20
        and "id" not in str(c).lower()
    ]

    # Prefer diagnosis_result when available
    group_col = next(
        (
            c for c in valid_cat_cols
            if str(c).strip().lower() == "diagnosis_result"
        ),
        valid_cat_cols[0] if valid_cat_cols else None
    )

    # Fallback to another categorical column
    if group_col is None:
        group_col = next(
            (
                c for c in categorical_cols
                if df[c].nunique(dropna=True) > 1
            ),
            None
        )

    # Step 4: Validate the selected columns
    if age_col is None or group_col is None:
        print(
            "Aggregation skipped: a suitable age or categorical "
            "column was not found."
        )
        print("Available columns:", df.columns.tolist())

    else:
        # Step 5: Prepare clean data
        agg_input = df[[group_col, age_col]].copy()

        agg_input[age_col] = pd.to_numeric(
            agg_input[age_col],
            errors="coerce"
        )

        agg_input[age_col] = agg_input[age_col].replace(
            [np.inf, -np.inf],
            np.nan
        )

        agg_input = agg_input.dropna(
            subset=[group_col, age_col]
        )

        if agg_input.empty:
            print("No valid observations are available for aggregation.")

        else:
            # Step 6: Calculate group-level statistics
            agg_df = (
                agg_input.groupby(group_col, observed=True)[age_col]
                .agg(["mean", "median", "count", "min", "max"])
                .reset_index()
            )

            agg_df.columns = [
                group_col,
                f"Mean_{age_col}",
                f"Median_{age_col}",
                "Patient_Count",
                f"Minimum_{age_col}",
                f"Maximum_{age_col}"
            ]

            agg_df = (
                agg_df.sort_values(
                    f"Mean_{age_col}",
                    ascending=False
                )
                .reset_index(drop=True)
            )

            # Step 7: Display the aggregation table
            print(f"Grouping column: {group_col}")
            print(f"Measurement column: {age_col}")
            print(f"Valid observations: {len(agg_input):,}")
            print(f"Number of groups: {len(agg_df):,}")

            display(agg_df.round(2))

            # Step 8: Create a static bar chart
            try:
                fig_width = max(
                    9,
                    min(16, len(agg_df) * 1.2)
                )

                fig, ax = plt.subplots(
                    figsize=(fig_width, 5.5)
                )

                bars = ax.bar(
                    agg_df[group_col].astype(str),
                    agg_df[f"Mean_{age_col}"],
                    edgecolor="black",
                    linewidth=0.5
                )

                ax.set_title(
                    f"Average {age_col} by {group_col}",
                    fontsize=14,
                    fontweight="bold",
                    pad=15
                )

                ax.set_xlabel(str(group_col), fontsize=11)
                ax.set_ylabel(f"Mean {age_col}", fontsize=11)

                ax.grid(
                    axis="y",
                    linestyle="--",
                    alpha=0.3
                )

                ax.set_axisbelow(True)

                # Add value labels above the bars
                for bar, value in zip(
                    bars,
                    agg_df[f"Mean_{age_col}"]
                ):
                    ax.annotate(
                        f"{value:.2f}",
                        (
                            bar.get_x() + bar.get_width() / 2,
                            bar.get_height()
                        ),
                        xytext=(0, 4),
                        textcoords="offset points",
                        ha="center",
                        va="bottom",
                        fontsize=9
                    )

                plt.xticks(rotation=30, ha="right")
                fig.tight_layout()
                plt.show()
                plt.close(fig)

            except Exception as plot_error:
                print("Chart rendering failed.")
                print("The aggregation table remains available.")
                print("Error details:", str(plot_error))

            # Step 9: Print a text-based summary
            print("\nGroup Mean Summary")
            print("-" * 45)

            for _, row in agg_df.iterrows():
                print(
                    f"{row[group_col]}: "
                    f"Mean age = {row[f'Mean_{age_col}']:.2f}, "
                    f"Median age = {row[f'Median_{age_col}']:.2f}, "
                    f"Patients = {int(row['Patient_Count'])}"
                )
```

## 26: Feature Importance Analysis
Extract feature importance weights from the trained Random Forest classifier and visualize top predictive drivers.

```python

# Top 10 Feature Importances for Classification — Kaggle-Safe

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

# Step 1: Display section heading
display(
    Markdown("## Top 10 Feature Importances for Classification")
)

# Step 2: Validate training data and pipeline
if "X_train" not in globals() or X_train is None:
    print("Feature importance analysis skipped: X_train is unavailable.")

elif "model_pipeline" not in globals() or model_pipeline is None:
    print("Feature importance analysis skipped: model_pipeline is unavailable.")

else:
    # Step 3: Identify the fitted classifier
    if hasattr(model_pipeline, "named_steps"):
        classifier = model_pipeline.named_steps.get("classifier")
    else:
        classifier = model_pipeline

    if classifier is None:
        print("No pipeline step named 'classifier' was found.")

    elif (
        not hasattr(classifier, "feature_importances_")
        and not hasattr(classifier, "coef_")
    ):
        print(
            f"{type(classifier).__name__} does not provide built-in "
            "feature_importances_ or coef_."
        )
        print(
            "Consider permutation importance for models without "
            "built-in feature importance."
        )

    else:
        # Step 4: Extract importance values
        if hasattr(classifier, "feature_importances_"):
            importance_values = np.asarray(
                classifier.feature_importances_,
                dtype=float
            ).ravel()

        else:
            # Absolute coefficients as an importance proxy
            coefficients = np.asarray(
                classifier.coef_,
                dtype=float
            )

            if coefficients.ndim == 1:
                importance_values = np.abs(coefficients)
            else:
                importance_values = np.mean(
                    np.abs(coefficients),
                    axis=0
                )

        # Step 5: Obtain original feature names
        if hasattr(X_train, "columns"):
            original_feature_names = [
                str(col) for col in X_train.columns
            ]
        else:
            X_array = np.asarray(X_train)

            if X_array.ndim == 1:
                n_original_features = 1
            else:
                n_original_features = X_array.shape[1]

            original_feature_names = [
                f"Feature_{i}"
                for i in range(n_original_features)
            ]

        feature_names = original_feature_names

        # Step 6: Try to obtain transformed feature names
        if hasattr(model_pipeline, "named_steps"):
            preprocessor = model_pipeline.named_steps.get(
                "preprocessor"
            )

            if (
                preprocessor is not None
                and hasattr(preprocessor, "get_feature_names_out")
            ):
                try:
                    transformed_names = list(
                        preprocessor.get_feature_names_out()
                    )

                    if len(transformed_names) == len(importance_values):
                        feature_names = [
                            str(name) for name in transformed_names
                        ]

                except Exception:
                    pass

        # Step 7: Validate feature names and values
        if len(importance_values) != len(feature_names):
            print("Feature names do not match importance values.")
            print(f"Importance values: {len(importance_values)}")
            print(f"Feature names: {len(feature_names)}")
            print(
                "The preprocessing pipeline may transform or expand "
                "the original features."
            )

        else:
            # Step 8: Create the feature importance table
            feat_imp = pd.DataFrame({
                "Feature": feature_names,
                "Importance": importance_values
            })

            feat_imp["Importance"] = pd.to_numeric(
                feat_imp["Importance"],
                errors="coerce"
            )

            feat_imp = feat_imp.replace(
                [np.inf, -np.inf],
                np.nan
            ).dropna(subset=["Importance"])

            feat_imp = (
                feat_imp.sort_values(
                    "Importance",
                    ascending=False
                )
                .head(10)
                .reset_index(drop=True)
            )

            if feat_imp.empty:
                print("No valid feature importance values were found.")

            else:
                # Display numerical results
                print(f"Classifier: {type(classifier).__name__}")
                print(f"Features displayed: {len(feat_imp)}")
                display(feat_imp.round(6))

                # Step 9: Create a static horizontal bar chart
                try:
                    plot_df = feat_imp.sort_values(
                        "Importance",
                        ascending=True
                    )

                    fig_height = max(
                        5,
                        min(10, len(plot_df) * 0.55)
                    )

                    fig, ax = plt.subplots(
                        figsize=(12, fig_height)
                    )

                    bars = ax.barh(
                        plot_df["Feature"].astype(str),
                        plot_df["Importance"],
                        edgecolor="black",
                        linewidth=0.4
                    )

                    ax.set_title(
                        "Top 10 Feature Importances for Classification",
                        fontsize=14,
                        fontweight="bold",
                        pad=15
                    )

                    ax.set_xlabel("Importance Score", fontsize=11)
                    ax.set_ylabel("Feature", fontsize=11)

                    ax.grid(
                        axis="x",
                        linestyle="--",
                        alpha=0.3
                    )

                    ax.set_axisbelow(True)

                    # Annotate each bar
                    min_value = plot_df["Importance"].min()
                    max_value = plot_df["Importance"].max()
                    value_range = max_value - min_value

                    offset = (
                        value_range * 0.015
                        if value_range > 0
                        else max(abs(max_value) * 0.02, 0.001)
                    )

                    for bar in bars:
                        value = bar.get_width()

                        if value >= 0:
                            x_position = value + offset
                            alignment = "left"
                        else:
                            x_position = value - offset
                            alignment = "right"

                        ax.annotate(
                            f"{value:.4f}",
                            (
                                x_position,
                                bar.get_y() + bar.get_height() / 2
                            ),
                            ha=alignment,
                            va="center",
                            fontsize=9
                        )

                    # Leave space for labels
                    left_limit = min(0, min_value - offset * 5)
                    right_limit = max(0, max_value + offset * 6)

                    if left_limit == right_limit:
                        right_limit = left_limit + 1

                    ax.set_xlim(left_limit, right_limit)

                    fig.tight_layout()
                    plt.show()
                    plt.close(fig)

                except Exception as plot_error:
                    print("Feature importance chart could not be rendered.")
                    print("The importance table remains available.")
                    print("Error details:", str(plot_error))
```
