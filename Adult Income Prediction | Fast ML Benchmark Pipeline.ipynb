```python
# ==========================================
# 1. Setup & Environment
# ==========================================
import warnings
warnings.filterwarnings('ignore')

import os
import glob
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Theme Setup
plt.style.use('dark_background')
plt.rcParams['figure.facecolor'] = '#0A0F1D'
plt.rcParams['axes.facecolor'] = '#121A2F'
plt.rcParams['axes.edgecolor'] = '#38BDF8'
plt.rcParams['axes.labelcolor'] = '#FFFFFF'
plt.rcParams['xtick.color'] = '#E2E8F0'
plt.rcParams['ytick.color'] = '#E2E8F0'
plt.rcParams['grid.color'] = '#1E293B'
plt.rcParams['grid.alpha'] = 0.5

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
print("Environment ready")
```

## 2. Data Ingestion

```python
# ==========================================
# 2. Data Ingestion
# ==========================================
DATASET_PATH = "/kaggle/input/datasets/serpilturanyksel/adult-income"
csv_files = glob.glob(os.path.join(DATASET_PATH, "*.csv")) + glob.glob("./*.csv") + glob.glob("/kaggle/input/*/*.csv")

if csv_files:
    data_file = csv_files[0]
    df = pd.read_csv(data_file)
    print(f"Dataset successfully loaded from: {data_file}")
else:
    raise FileNotFoundError("Dataset path not found. Please attach the dataset to your Kaggle environment.")

# Clean up column names and string whitespace
df.columns = df.columns.str.strip().str.lower().str.replace('-', '_').str.replace('.', '_')
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str).str.strip()

df.replace('?', np.nan, inplace=True)
print(f"Data Matrix Dimensions: {df.shape[0]:,} Rows | {df.shape[1]} Columns")

```

## 3. Target Identification & Quick EDA

```python
# ==========================================
# 3. Target Identification & Quick EDA
# ==========================================
target_col = [c for c in df.columns if 'income' in c or 'target' in c or 'class' in c][0]
df[target_col] = df[target_col].str.replace('.', '', regex=False)

fig, ax = plt.subplots(figsize=(6, 3.5))
sns.countplot(data=df, x=target_col, palette=['#38BDF8', '#00E5FF'], ax=ax)
ax.set_title(f"Income Distribution ({target_col})", fontsize=12, fontweight='bold', color='#38BDF8')
ax.set_xlabel("Income Class", fontsize=10)
ax.set_ylabel("Count", fontsize=10)
plt.tight_layout()
plt.show()
```

## 4. Preprocessing & Memory-Safe Split

```python
# ==========================================
# 4. Preprocessing & Memory-Safe Split
# ==========================================
X = df.drop(columns=[target_col])
y = (df[target_col].isin(['>50K', '>50K.'])).astype(int)

# Drop redundant or zero-variance columns if present
fnlwgt_col = [c for c in X.columns if 'fnlwgt' in c]
if fnlwgt_col:
    X = X.drop(columns=fnlwgt_col)

num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=True), cat_cols)
    ]
)
```

## 5. Fast Model Benchmark

```python
# ==========================================
# 5. Fast Model Benchmark
# ==========================================
models = {
    "Hist Gradient Boosting": HistGradientBoostingClassifier(random_state=RANDOM_STATE),
    "Random Forest": RandomForestClassifier(n_estimators=100, max_depth=12, random_state=RANDOM_STATE, n_jobs=2)
}

benchmark_results = []

for name, model in models.items():
    if name == "Hist Gradient Boosting":
        # Convert sparse output to dense safely for HistGradientBoosting
        X_tr_proc = preprocessor.fit_transform(X_train)
        X_te_proc = preprocessor.transform(X_test)
        if hasattr(X_tr_proc, "toarray"):
            X_tr_proc = X_tr_proc.toarray()
            X_te_proc = X_te_proc.toarray()
        
        model.fit(X_tr_proc, y_train)
        preds = model.predict(X_te_proc)
    else:
        pipe = Pipeline([('preprocessor', preprocessor), ('classifier', model)])
        pipe.fit(X_train, y_train)
        preds = pipe.predict(X_test)
    
    benchmark_results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, preds),
        'Precision': precision_score(y_test, preds),
        'Recall': recall_score(y_test, preds),
        'F1 Score': f1_score(y_test, preds)
    })

benchmark_df = pd.DataFrame(benchmark_results).sort_values(by='F1 Score', ascending=False)
print("=== Model Benchmark Summary ===")
display(benchmark_df)
```

## 6. Winning Pipeline & Confusion Matrix

```python

# ==========================================
# 6. Winning Pipeline & Confusion Matrix
# ==========================================
winning_model = HistGradientBoostingClassifier(random_state=RANDOM_STATE)
X_proc_all = preprocessor.fit_transform(X)
if hasattr(X_proc_all, "toarray"):
    X_proc_all = X_proc_all.toarray()

X_tr_p, X_te_p, y_tr, y_te = train_test_split(X_proc_all, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y)
winning_model.fit(X_tr_p, y_tr)
winning_preds = winning_model.predict(X_te_p)

fig, ax = plt.subplots(figsize=(5, 4))
cm = confusion_matrix(y_te, winning_preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax, cbar=False,
            xticklabels=['<=50K', '>50K'], yticklabels=['<=50K', '>50K'])
ax.set_title("Confusion Matrix (Hist Gradient Boosting)", fontsize=11, fontweight='bold', color='#38BDF8')
ax.set_xlabel("Predicted Class")
ax.set_ylabel("True Class")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 7. Interactive Income Predictor Widget
# ==========================================
import ipywidgets as widgets
from IPython.display import display, HTML

feature_names = X.columns.tolist()
input_widgets = {}

for col in feature_names:
    if np.issubdtype(X[col].dtype, np.number):
        min_v, max_v, mean_v = float(X[col].min()), float(X[col].max()), float(X[col].mean())
        input_widgets[col] = widgets.FloatSlider(value=mean_v, min=min_v, max=max_v, step=(max_v - min_v)/100, description=f"{col}:")
    else:
        unique_vals = [v for v in X[col].dropna().unique().tolist() if v != 'nan']
        input_widgets[col] = widgets.Dropdown(options=unique_vals, value=unique_vals[0], description=f"{col}:")

predict_btn = widgets.Button(description="Predict Income", button_style='primary', layout=widgets.Layout(width='200px'))
output_box = widgets.Output()

def on_click(b):
    with output_box:
        output_box.clear_output()
        input_df = pd.DataFrame([{c: input_widgets[c].value for c in feature_names}])
        proc_input = preprocessor.transform(input_df)
        if hasattr(proc_input, "toarray"):
            proc_input = proc_input.toarray()
        pred = winning_model.predict(proc_input)[0]
        status = ">50K" if pred == 1 else "<=50K"
        color = "#38BDF8" if pred == 1 else "#FF1744"
        display(HTML(f"""
        <div style="background:#121A2F; border:1px solid {color}; padding:12px; border-radius:8px; margin-top:10px; width:260px; text-align:center;">
            <span style="color:#94A3B8; font-size:11px;">PREDICTED INCOME BRACKET</span>
            <h2 style="color:{color}; font-size:24px; margin:4px 0 0 0;">{status}</h2>
        </div>
        """))

predict_btn.on_click(on_click)
display(HTML("<h4 style='color:#38BDF8;'>Interactive Income Profile Controls</h4>"))
display(widgets.GridBox(list(input_widgets.values()), layout=widgets.Layout(grid_template_columns='repeat(2, 300px)', grid_gap='10px')))
display(predict_btn)
display(output_box)
```

## 8. Save Artifacts & Generate Submission CSV

```python

# ==========================================
# 8. Save Artifacts & Generate Submission CSV
# ==========================================
submission_df = pd.DataFrame({
    'Id': np.arange(1, len(winning_preds) + 1),
    'Predicted_Income': np.where(winning_preds == 1, '>50K', '<=50K')
})
submission_path = "/kaggle/working/adult_income_predictions.csv"
submission_df.to_csv(submission_path, index=False)

model_path = "/kaggle/working/adult_income_model.joblib"
joblib.dump(winning_model, model_path)

print(f"Predictions exported to: {submission_path}")
print(f"Model saved to: {model_path}")
```
