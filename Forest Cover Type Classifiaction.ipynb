## 1. Import Libraries
In this step, we import essential numerical, data manipulation, visualization, and machine learning libraries.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_covtype
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

print("All dependencies loaded successfully.")
```

## 2. Set Random Seed
 Setting a reproducible random seed ensures consistent outputs across runs.

```python
import warnings
warnings.filterwarnings('ignore')

SEED = 42
np.random.seed(SEED)

print("Random Seed Setted")
```

## 3. Load the UCI Forest Covertype Dataset
We load the Covertype dataset directly using Scikit-Learn's dataset fetcher. This dataset consists of 581,012 samples and 54 features describing cartographic variables for $30 \times 30$ meter cells.

```python
# Fetch dataset (downloads automatically if not cached)
covtype_data = fetch_covtype(as_frame=True)

# Extract features and target variable
X = covtype_data.data
y = covtype_data.target - 1  # Convert labels (1-7) to zero-indexed (0-6) for XGBoost compatibility

df = X.copy()
df['Cover_Type'] = y

print(f"Dataset shape: {df.shape}")
```

## 4. Inspect Dataset Overview
Let's inspect the first few rows of the dataframe to understand the structure of the cartographic features, Wilderness Areas, and Soil Types.

```python
df.head()
```

## 5. Exploratory Data Analysis (EDA) - Missing Values & Class Distribution
We check for missing values and examine the multi-class target variable distribution (Cover_Type classes 0 through 6).

```python
print("--- Missing Values ---")
print(f"Total Missing Values: {df.isnull().sum().sum()}")

print("\n--- Target Class Distribution ---")
class_counts = df['Cover_Type'].value_counts().sort_index()
print(class_counts)

plt.figure(figsize=(8, 4))
sns.barplot(x=class_counts.index + 1, y=class_counts.values, palette='viridis')
plt.title('Distribution of Forest Cover Types (Classes 1-7)')
plt.xlabel('Cover Type Class')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

## 6. Summary Statistics of Continuous Features
The first 10 columns are continuous environmental features (e.g., Elevation, Aspect, Slope, Distances to Hydrology/Roadways/Fire Points). Let's review their statistics.

```python
continuous_cols = [
    'Elevation', 'Aspect', 'Slope', 
    'Horizontal_Distance_To_Hydrology', 'Vertical_Distance_To_Hydrology',
    'Horizontal_Distance_To_Roadways', 'Hillshade_9am', 'Hillshade_Noon',
    'Hillshade_3pm', 'Horizontal_Distance_To_Fire_Points'
]

df[continuous_cols].describe().T
```

## 7. Feature Preprocessing & Scaling
Because the dataset is large (581,012 rows), sub-sampling or feature scaling helps standard models converge faster. We standard scale the continuous variables while keeping one-hot binary columns intact.

```python
scaler = StandardScaler()
X_scaled = X.copy()
X_scaled[continuous_cols] = scaler.fit_transform(X[continuous_cols])

print("Continuous features successfully scaled.")
X_scaled[continuous_cols].head(3)
```

## 8. Subsampling for Rapid Prototyping
To make training efficient during comparison and tuning, we draw a stratified sample of 100,000 instances from the full dataset.

```python
X_sample, _, y_sample, _ = train_test_split(
    X_scaled, y, 
    train_size=100000, 
    stratify=y, 
    random_state=SEED
)

print(f"Sampled dataset size: {X_sample.shape[0]} rows")
```

## 9. Stratified Train-Test Split
We split the sampled dataset into 80% training and 20% validation/testing subsets, keeping class distributions proportional.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_sample, y_sample, 
    test_size=0.2, 
    stratify=y_sample, 
    random_state=SEED
)

print(f"Training Set: {X_train.shape[0]} samples")
print(f"Testing Set:  {X_test.shape[0]} samples")
```

## 10. Model 1: Baseline Decision Tree Classifier
We establish a baseline using a standard single Decision Tree model.

```python
dt_model = DecisionTreeClassifier(random_state=SEED, max_depth=15)
dt_model.fit(X_train, y_train)

dt_preds = dt_model.predict(X_test)
dt_acc = accuracy_score(y_test, dt_preds)
print(f"Decision Tree Accuracy: {dt_acc * 100:.2f}%")
```

## 11. Model 2: Random Forest Classifier
Next, we train an ensemble model—Random Forest—which leverages multiple bootstrapped decision trees to reduce variance and improve multi-class accuracy.

```python
rf_model = RandomForestClassifier(n_estimators=100, random_state=SEED, n_jobs=-1)
rf_model.fit(X_train, y_train)

rf_preds = rf_model.predict(X_test)
rf_acc = accuracy_score(y_test, rf_preds)
print(f"Random Forest Accuracy: {rf_acc * 100:.2f}%")
```

## 12. Model 3: XGBoost Classifier
We train a Gradient Boosting model using XGBoost with multi-class softmax loss.

```python
xgb_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=8,
    objective='multi:softprob',
    num_class=7,
    random_state=SEED,
    n_jobs=-1
)

xgb_model.fit(X_train, y_train)

xgb_preds = xgb_model.predict(X_test)
xgb_acc = accuracy_score(y_test, xgb_preds)
print(f"XGBoost Accuracy: {xgb_acc * 100:.2f}%")
```

## 13. Model Performance Comparison
We compare the performance metrics across all three trained classifiers.

```python
model_results = pd.DataFrame({
    'Model': ['Decision Tree', 'Random Forest', 'XGBoost'],
    'Accuracy (%)': [dt_acc * 100, rf_acc * 100, xgb_acc * 100]
})

print(model_results)

plt.figure(figsize=(7, 4))
sns.barplot(x='Model', y='Accuracy (%)', data=model_results, palette='crest')
plt.ylim(50, 100)
plt.title('Multi-Class Model Accuracy Comparison')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

## 14. Detailed Classification Report (Best Model)
Let's examine the class-wise Precision, Recall, and F1-Scores for the best-performing model (XGBoost / Random Forest).

```python
best_preds = xgb_preds if xgb_acc >= rf_acc else rf_preds
best_model_name = "XGBoost" if xgb_acc >= rf_acc else "Random Forest"

print(f"--- Detailed Classification Report: {best_model_name} ---")
print(classification_report(y_test, best_preds, target_names=[f"Cover_Type_{i+1}" for i in range(7)]))
```

## 15. Confusion Matrix Visualization
A confusion matrix visualizes how accurately each target cover class is being predicted and highlights common misclassifications across target classes.

```python
cm = confusion_matrix(y_test, best_preds)

plt.figure(figsize=(9, 7))
sns.heatmap(
    cm, 
    annot=True, 
    fmt='d', 
    cmap='Blues', 
    xticklabels=[f"Type {i+1}" for i in range(7)],
    yticklabels=[f"Type {i+1}" for i in range(7)]
)
plt.title(f'Confusion Matrix - {best_model_name}')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

## 16. Feature Importance Extraction
We extract feature importances from our best ensemble model to identify which cartographic features contribute most to predicting cover types.

```python
best_model = xgb_model if xgb_acc >= rf_acc else rf_model

importances = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': best_model.feature_importances_
}).sort_values('Importance', ascending=False)

importances.head(15)
```

## 17. Top 15 Feature Importances Visualization
We plot the top 15 most influential predictors (e.g., Elevation, Distance to Roadways, Wilderness Areas).

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Feature', data=importances.head(15), palette='mako')
plt.title(f'Top 15 Most Important Features ({best_model_name})')
plt.xlabel('Feature Importance Score')
plt.ylabel('Feature Name')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

## 18. Hyperparameter Tuning via GridSearchCV
We perform Grid Search on Random Forest hyperparameters (max_depth, n_estimators) to optimize classification efficiency and prevent overfitting.

```python
param_grid = {
    'n_estimators': [50, 100],
    'max_depth': [10, 20],
    'min_samples_split': [2, 5]
}

rf_grid = RandomForestClassifier(random_state=SEED, n_jobs=-1)
grid_search = GridSearchCV(
    estimator=rf_grid, 
    param_grid=param_grid, 
    cv=3, 
    scoring='accuracy', 
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print(f"Best Hyperparameters: {grid_search.best_params_}")
print(f"Best Grid Search Score: {grid_search.best_score_ * 100:.2f}%")
```

## 19. Final Model Evaluation & Conclusion
We evaluate the final tuned model on the unseen test dataset to confirm generalized performance.

```python
tuned_model = grid_search.best_estimator_
tuned_preds = tuned_model.predict(X_test)

final_accuracy = accuracy_score(y_test, tuned_preds)
print(f"Final Tuned Random Forest Accuracy: {final_accuracy * 100:.2f}%")
```
