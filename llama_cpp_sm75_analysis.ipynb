## 3. Identify dataset directory

This section performs **identify dataset directory** as part of the complete software-dataset audit.

```python
slug = "llama-cpp-turing-sm75"
matching = [p for p in input_root.rglob("*") if p.is_dir() and slug in p.name.lower()]
if matching:
    DATA_DIR = matching[0]
else:
    scores = {}
    for p in all_files:
        score = sum(x in str(p.parent).lower() for x in ["llama", "cpp", "turing", "sm75"])
        scores[p.parent] = score
    if not scores:
        raise FileNotFoundError("No files found under /kaggle/input. Add the dataset first.")
    DATA_DIR = max(scores, key=scores.get)
print("Dataset directory:", DATA_DIR)
```

## 9. Load JSON metadata

This section performs **load json metadata** as part of the complete software-dataset audit.

```python
json_objects = {}
for path in json_files:
    try:
        with path.open("r", encoding="utf-8") as f:
            obj = json.load(f)
        json_objects[str(path.relative_to(DATA_DIR))] = obj
        print(path.name, "->", type(obj).__name__)
    except Exception as e:
        print("Could not parse", path.name, ":", e)
```

## 10. Summarize JSON structures

This section performs **summarize json structures** as part of the complete software-dataset audit.

```python
def json_shape(obj, depth=0, max_depth=3):
    if depth > max_depth:
        return "..."
    if isinstance(obj, dict):
        return {str(k): json_shape(v, depth+1, max_depth) for k,v in list(obj.items())[:25]}
    if isinstance(obj, list):
        return {"type":"list", "length":len(obj),
                "sample": json_shape(obj[0], depth+1, max_depth) if obj else None}
    return type(obj).__name__

for name, obj in json_objects.items():
    print("\n===", name, "===")
    display(json_shape(obj))
```

## 11. Search JSON for SM75/Turing references

This section performs **search json for sm75/turing references** as part of the complete software-dataset audit.

```python
def flatten_text(obj):
    if isinstance(obj, dict):
        for k,v in obj.items():
            yield str(k)
            yield from flatten_text(v)
    elif isinstance(obj, list):
        for x in obj:
            yield from flatten_text(x)
    else:
        yield str(obj)

rows = []
for name,obj in json_objects.items():
    text = " ".join(flatten_text(obj))
    hits = re.findall(r"sm[_-]?75|turing", text, flags=re.I)
    rows.append({"json_file":name, "sm75_turing_mentions":len(hits), "text_length":len(text)})
display(pd.DataFrame(rows))
```

## 17. Generate SHA-256 checksums

This section performs **generate sha-256 checksums** as part of the complete software-dataset audit.

```python
def sha256_file(path, chunk_size=1024*1024):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk: break
            h.update(chunk)
    return h.hexdigest()

checksums = pd.DataFrame([
    {"file":p.name, "sha256":sha256_file(p), "size_bytes":p.stat().st_size}
    for p in dataset_files
])
display(checksums)
```

## 21. Extract wheel METADATA

This section performs **extract wheel metadata** as part of the complete software-dataset audit.

```python
# Extract package metadata from ZIP/WHEEL artifacts

rows = []

for p in dataset_files:
    if not zipfile.is_zipfile(p):
        continue

    try:
        with zipfile.ZipFile(p, "r") as z:
            metadata_files = [
                name for name in z.namelist()
                if name.upper().endswith("/METADATA")
            ]

            if not metadata_files:
                print(f"No METADATA found: {p.name}")
                continue

            for metadata_file in metadata_files:
                text = z.read(metadata_file).decode(
                    "utf-8",
                    errors="replace"
                )

                fields = {
                    "Name": "",
                    "Version": "",
                    "Requires-Python": "",
                    "Platform": "",
                    "Summary": ""
                }

                for line in text.splitlines():
                    if ":" in line:
                        key, value = line.split(":", 1)
                        key = key.strip()
                        value = value.strip()

                        if key in fields:
                            fields[key] = value

                rows.append({
                    "Artifact": p.name,
                    "Metadata_File": metadata_file,
                    **fields
                })

    except Exception as e:
        print(f"Metadata error: {p.name} -> {e}")

metadata_df = pd.DataFrame(rows)

if metadata_df.empty:
    print("No package METADATA was found inside the ZIP/WHEEL artifacts.")
else:
    print(f"Package metadata found for {len(metadata_df)} artifact(s).")
    display(metadata_df)
```

## 23. Inspect RECORD entries

This section performs **inspect record entries** as part of the complete software-dataset audit.

```python
# Inspect RECORD files inside ZIP/WHEEL artifacts

rows = []

for p in dataset_files:
    if not zipfile.is_zipfile(p):
        continue

    try:
        with zipfile.ZipFile(p, "r") as z:
            record_files = [
                name for name in z.namelist()
                if name.upper().endswith("/RECORD")
            ]

            if not record_files:
                rows.append({
                    "artifact": p.name,
                    "record_found": False,
                    "packaged_entries": 0
                })
                continue

            total_entries = 0

            for record_file in record_files:
                try:
                    text = z.read(record_file).decode(
                        "utf-8",
                        errors="replace"
                    )

                    entries = [
                        line for line in text.splitlines()
                        if line.strip()
                    ]

                    total_entries += len(entries)

                except Exception as e:
                    print(f"Could not read RECORD in {p.name}: {e}")

            rows.append({
                "artifact": p.name,
                "record_found": True,
                "record_files": len(record_files),
                "packaged_entries": total_entries
            })

    except Exception as e:
        print(f"Archive error: {p.name} -> {e}")

record_df = pd.DataFrame(rows)

if record_df.empty:
    print("No ZIP/WHEEL artifacts with RECORD files were found.")
else:
    print(f"Analyzed {len(record_df)} archive artifact(s).")
    display(record_df)
```

## 27. Search text-readable files for CUDA/SM75

```python
hits = []
for p in dataset_files:
    if p.suffix.lower() in {".json",".txt",".md",".toml",".cfg",".ini"}:
        try:
            text = p.read_text(encoding="utf-8", errors="ignore")
            for term in ["sm75","sm_75","turing","cuda"]:
                count = len(re.findall(term, text, flags=re.I))
                if count:
                    hits.append({"file":p.name,"term":term,"occurrences":count})
        except Exception:
            pass
display(pd.DataFrame(hits))
```

## 28. Build compatibility table
