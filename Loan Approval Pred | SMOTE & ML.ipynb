## 1. Import Libraries and Dependencies
We begin by importing necessary data manipulation libraries, visualization packages, evaluation metrics, standard scikit-learn models, and imblearn for SMOTE resampling.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, precision_recall_curve
)
from imblearn.over_sampling import SMOTE

import warnings
warnings.filterwarnings('ignore')

SEED = 42
np.random.seed(SEED)
print("Environment initialized successfully.")
```

## 2. Dataset Loading and Initial Inspection
We load the dataset (or generate a standard synthetic representation matching Kaggle's Loan Approval schema if running offline) and inspect column structures.

```python
# Try loading Kaggle dataset path; if not found, load/generate standard loan dataset schema
try:
    df = pd.read_csv('/kaggle/input/datasets/architsharma01/loan-approval-prediction-dataset/loan_approval_dataset.csv')
except FileNotFoundError:
    try:
        df = pd.read_csv('loan_data.csv')
    except FileNotFoundError:
        # Generate representative Loan dataset structure for standalone execution
        n = 1000
        df = pd.DataFrame({
            'Loan_ID': [f'LP{i:05d}' for i in range(n)],
            'Gender': np.random.choice(['Male', 'Female', np.nan], size=n, p=[0.75, 0.20, 0.05]),
            'Married': np.random.choice(['Yes', 'No', np.nan], size=n, p=[0.60, 0.37, 0.03]),
            'Dependents': np.random.choice(['0', '1', '2', '3+', np.nan], size=n, p=[0.55, 0.18, 0.17, 0.07, 0.03]),
            'Education': np.random.choice(['Graduate', 'Not Graduate'], size=n, p=[0.78, 0.22]),
            'Self_Employed': np.random.choice(['Yes', 'No', np.nan], size=n, p=[0.12, 0.83, 0.05]),
            'ApplicantIncome': np.random.exponential(scale=5000, size=n).astype(int) + 1500,
            'CoapplicantIncome': np.random.exponential(scale=2000, size=n).astype(int),
            'LoanAmount': np.random.normal(loc=140, scale=50, size=n).clip(20, 500),
            'Loan_Amount_Term': np.random.choice([360, 180, 240, 120, np.nan], size=n, p=[0.8, 0.1, 0.05, 0.02, 0.03]),
            'Credit_History': np.random.choice([1.0, 0.0, np.nan], size=n, p=[0.75, 0.18, 0.07]),
            'Property_Area': np.random.choice(['Urban', 'Semiurban', 'Rural'], size=n, p=[0.33, 0.37, 0.30]),
            'Loan_Status': np.random.choice(['Y', 'N'], size=n, p=[0.69, 0.31])
        })

print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
# Clean column names by removing extra whitespace
df.columns = df.columns.str.strip()

# Print column names to confirm exact casing
print("Dataset Columns:", df.columns.tolist())
```

## 3. Exploratory Data Analysis (EDA) - Missing Values & Class Imbalance
We examine missing values per column and evaluate the class distribution of Loan_Status to check for target class imbalance.

```python
print("--- Missing Values Count ---")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\n--- Target Distribution (loan_status) ---")
status_counts = df['loan_status'].value_counts()
print(status_counts)

plt.figure(figsize=(6, 4))
sns.countplot(x='loan_status', data=df, palette='Set2')
plt.title('Target Variable Distribution (loan_status)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

## 4. Handling Missing Values
Categorical columns with missing values are imputed with their mode, while continuous numerical variables are imputed with their median.

```python
# Identify numerical and categorical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

# Fill numerical missing values with median
for col in num_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].median(), inplace=True)

# Fill categorical missing values with mode
for col in cat_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].mode()[0], inplace=True)

print(f"Remaining Missing Values: {df.isnull().sum().sum()}")
```

## 5. Categorical Encoding & Feature Engineering
We convert categorical string features into numerical representations using One-Hot Encoding (pd.get_dummies) and binary mapping for Loan_Status.

```python
# Drop non-predictive ID column
if 'loan_id' in df.columns:
    df.drop('loan_id', axis=1, inplace=True)

# Clean string values and map target column to binary (1 / 0)
df['loan_status'] = df['loan_status'].astype(str).str.strip()
df['loan_status'] = df['loan_status'].map({'Approved': 1, 'Rejected': 0, 'Y': 1, 'N': 0})

# Clean string whitespace for all remaining object columns
categorical_features = df.select_dtypes(include=['object']).columns
for col in categorical_features:
    df[col] = df[col].astype(str).str.strip()

# One-hot encode categorical features
df = pd.get_dummies(df, columns=categorical_features, drop_first=True)

print(f"Processed Feature Set Shape: {df.shape}")
df.head(3)
```

## 6. Stratified Train-Test Split
We separate target labels from feature vectors and perform an 80/20 train-test split using stratified sampling.

```python
X = df.drop('loan_status', axis=1)
y = df['loan_status']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=SEED
)

print(f"Train Set Shape: {X_train.shape}")
print(f"Test Set Shape:  {X_test.shape}")
```

## 7. Feature Scaling
We scale continuous features using StandardScaler to normalize feature magnitudes for model training.

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling successfully applied.")
```

## 8. Addressing Class Imbalance with SMOTE
We apply Synthetic Minority Over-sampling Technique (SMOTE) on the training features to balance minority class samples.

```python
print(f"Class counts BEFORE SMOTE: {np.bincount(y_train)}")

smote = SMOTE(random_state=SEED)
X_train_res, y_train_res = smote.fit_resample(X_train_scaled, y_train)

print(f"Class counts AFTER SMOTE:  {np.bincount(y_train_res)}")
```

## 9. Model 1: Logistic Regression
We train a Logistic Regression classifier on the SMOTE-balanced training dataset.

```python
lr_model = LogisticRegression(random_state=SEED, max_iter=1000)
lr_model.fit(X_train_res, y_train_res)

lr_preds = lr_model.predict(X_test_scaled)
print("Logistic Regression Model Trained Successfully.")
```

## 10. Model 2: Decision Tree Classifier
We train a Decision Tree classifier, setting max_depth to mitigate overfitting on synthetic samples.

```python
dt_model = DecisionTreeClassifier(max_depth=5, random_state=SEED)
dt_model.fit(X_train_res, y_train_res)

dt_preds = dt_model.predict(X_test_scaled)
print("Decision Tree Model Trained Successfully.")
```

## 11. Comparative Model Evaluation (Focusing on Precision, Recall, F1-Score)
We evaluate both models using precision, recall, and F1-score to analyze their performance on imbalanced loan approvals.

```python
def get_metrics(y_true, y_pred, model_name):
    return {
        'Model': model_name,
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred),
        'Recall': recall_score(y_true, y_pred),
        'F1-Score': f1_score(y_true, y_pred)
    }

results_df = pd.DataFrame([
    get_metrics(y_test, lr_preds, 'Logistic Regression'),
    get_metrics(y_test, dt_preds, 'Decision Tree')
])

print("=== Evaluation Summary ===")
print(results_df.to_string(index=False))
```

## 12. Confusion Matrix Visualizations
Visualizing side-by-side confusion matrices highlights false positives and false negatives for both models.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

sns.heatmap(confusion_matrix(y_test, lr_preds), annot=True, fmt='d', cmap='Blues', ax=axes[0])
axes[0].set_title('Logistic Regression Confusion Matrix')
axes[0].set_xlabel('Predicted')
axes[0].set_ylabel('Actual')

sns.heatmap(confusion_matrix(y_test, dt_preds), annot=True, fmt='d', cmap='Greens', ax=axes[1])
axes[1].set_title('Decision Tree Confusion Matrix')
axes[1].set_xlabel('Predicted')
axes[1].set_ylabel('Actual')

plt.tight_layout()
plt.show()
```

## 13. Precision-Recall Curve Analysis
Given class imbalance, Precision-Recall curves provide a clearer picture of classifier quality than ROC curves.

```python
lr_probs = lr_model.predict_proba(X_test_scaled)[:, 1]
dt_probs = dt_model.predict_proba(X_test_scaled)[:, 1]

p_lr, r_lr, _ = precision_recall_curve(y_test, lr_probs)
p_dt, r_dt, _ = precision_recall_curve(y_test, dt_probs)

plt.figure(figsize=(7, 5))
plt.plot(r_lr, p_lr, label='Logistic Regression', color='blue')
plt.plot(r_dt, p_dt, label='Decision Tree', color='green')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve Comparison')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## 14. Feature Importance Analysis
We extract feature importances from the Decision Tree model to understand which variables drive loan approval decisions most strongly.

```python
importances = pd.DataFrame({
    'Feature': X.columns,
    'Importance': dt_model.feature_importances_
}).sort_values('Importance', ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x='Importance', y='Feature', data=importances.head(10), palette='crest')
plt.title('Top 10 Decision Tree Feature Importances')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

## 15. Conclusion and Summary
We print the final comparative summary of classification metrics and key takeaways for loan risk modeling.

```python
print("--- Final Model Evaluation Report ---")
print("\nLogistic Regression Detailed Report:")
print(classification_report(y_test, lr_preds, target_names=['Denied (0)', 'Approved (1)']))

print("\nDecision Tree Detailed Report:")
print(classification_report(y_test, dt_preds, target_names=['Denied (0)', 'Approved (1)']))
```
