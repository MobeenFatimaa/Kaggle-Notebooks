```python
# ==========================================
# 1. Setup & Environment
# ==========================================
import warnings
warnings.filterwarnings('ignore')

import os
import glob
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# TikTok Cyan-Pink Dark Theme Setup
plt.style.use('dark_background')
plt.rcParams['figure.facecolor'] = '#0D0221'
plt.rcParams['axes.facecolor'] = '#151D30'
plt.rcParams['axes.edgecolor'] = '#FE2C55'
plt.rcParams['axes.labelcolor'] = '#FFFFFF'
plt.rcParams['xtick.color'] = '#E2E8F0'
plt.rcParams['ytick.color'] = '#E2E8F0'
plt.rcParams['grid.color'] = '#252D42'
plt.rcParams['grid.alpha'] = 0.5

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
print("Environment Ready")
```

## 2. Data Ingestion

```python
# ==========================================
# 2. Data Ingestion
# ==========================================
DATASET_PATH = "/kaggle/input/datasets/datascikhan/tiktok-shop-viral-sales-and-cac-attribution"
csv_files = glob.glob(os.path.join(DATASET_PATH, "*.csv")) + glob.glob("./*.csv") + glob.glob("/kaggle/input/*/*.csv")

if csv_files:
    data_file = csv_files[0]
    df = pd.read_csv(data_file)
    print(f"Dataset successfully loaded from: {data_file}")
else:
    raise FileNotFoundError("Dataset path not found. Please attach the dataset to your Kaggle environment.")

# Standardize column headers
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_')
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str).str.strip()

print(f"Data Dimensions: {df.shape[0]:,} Rows | {df.shape[1]} Columns")

# Target Engineering (High Virality / High Conversion Indicator)
target_candidates = [c for c in df.columns if 'viral' in c or 'converted' in c or 'target' in c or 'class' in c]
if target_candidates:
    target_col = target_candidates[0]
else:
    # Synthetic threshold baseline if continuous
    num_cols_temp = df.select_dtypes(include=[np.number]).columns
    target_col = num_cols_temp[-1]
    df[target_col] = (df[target_col] > df[target_col].median()).astype(int)
```

## 3. Exploratory Data Analysis

```python
# ==========================================
# 3. Exploratory Data Analysis
# ==========================================
fig, ax = plt.subplots(figsize=(6, 3.5))
sns.countplot(data=df, x=target_col, palette=['#FE2C55', '#25F4EE'], ax=ax)
ax.set_title(f"Class Distribution: {target_col}", fontsize=12, fontweight='bold', color='#25F4EE')
ax.set_xlabel("Virality Target Class", fontsize=10)
ax.set_ylabel("Count", fontsize=10)
plt.tight_layout()
plt.show()
```

## 4. Preprocessing & Safe Train-Test Split

```python
# ==========================================
# 4. Preprocessing & Safe Train-Test Split
# ==========================================
from sklearn.preprocessing import OrdinalEncoder

X = df.drop(columns=[target_col])
y = df[target_col].astype(int)

# Drop explicit ID or high-cardinality string columns (e.g., unique codes, names, dates)
id_cols = [c for c in X.columns if 'id' in c.lower() or 'code' in c.lower() or 'url' in c.lower() or 'date' in c.lower()]
if id_cols:
    X = X.drop(columns=id_cols)

num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

# Use OrdinalEncoder instead of OneHotEncoder to prevent column explosion and RAM crashes
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), cat_cols)
    ]
)

# Preprocess directly to memory-efficient float32 dense arrays
X_tr_proc = preprocessor.fit_transform(X_train).astype(np.float32)
X_te_proc = preprocessor.transform(X_test).astype(np.float32)
```

## 5. Fast Model Benchmark (Memory-Safe)

```python
# ==========================================
# 5. Fast Model Benchmark (Zero-Crash Edition)
# ==========================================
models = {
    "Hist Gradient Boosting": HistGradientBoostingClassifier(random_state=RANDOM_STATE),
    "Random Forest": RandomForestClassifier(n_estimators=100, max_depth=10, random_state=RANDOM_STATE, n_jobs=2)
}

benchmark_results = []

for name, model in models.items():
    model.fit(X_tr_proc, y_train)
    preds = model.predict(X_te_proc)
    
    benchmark_results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, preds),
        'Precision': precision_score(y_test, preds, average='macro'),
        'Recall': recall_score(y_test, preds, average='macro'),
        'F1 Score': f1_score(y_test, preds, average='macro')
    })

benchmark_df = pd.DataFrame(benchmark_results).sort_values(by='F1 Score', ascending=False)
print("=== TikTok Attribution Model Benchmark ===")
display(benchmark_df)
```

## 6. Evaluation & Confusion Matrix

```python
# ==========================================
# 6. Evaluation & Confusion Matrix
# ==========================================
winning_model = HistGradientBoostingClassifier(random_state=RANDOM_STATE)
winning_model.fit(X_tr_proc, y_train)
winning_preds = winning_model.predict(X_te_proc)

fig, ax = plt.subplots(figsize=(5, 4))
cm = confusion_matrix(y_test, winning_preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='RdPu', ax=ax, cbar=False)
ax.set_title("Confusion Matrix (Hist Gradient Boosting)", fontsize=11, fontweight='bold', color='#25F4EE')
ax.set_xlabel("Predicted Class")
ax.set_ylabel("True Class")
plt.tight_layout()
plt.show()
```

## 7. Interactive Virality & CAC Predictor

```python
# ==========================================
# 7. Interactive Virality & CAC Predictor
# ==========================================
import ipywidgets as widgets
from IPython.display import display, HTML

feature_names = X.columns.tolist()
input_widgets = {}

for col in feature_names:
    if np.issubdtype(X[col].dtype, np.number):
        min_v, max_v, mean_v = float(X[col].min()), float(X[col].max()), float(X[col].mean())
        step_v = (max_v - min_v)/100 if max_v != min_v else 1.0
        input_widgets[col] = widgets.FloatSlider(value=mean_v, min=min_v, max=max_v, step=step_v, description=f"{col[:15]}:")
    else:
        unique_vals = [v for v in X[col].dropna().unique().tolist() if v != 'nan']
        input_widgets[col] = widgets.Dropdown(options=unique_vals, value=unique_vals[0], description=f"{col[:15]}:")

predict_btn = widgets.Button(description="Predict Virality", button_style='danger', layout=widgets.Layout(width='200px'))
output_box = widgets.Output()

def on_click(b):
    with output_box:
        output_box.clear_output()
        input_df = pd.DataFrame([{c: input_widgets[c].value for c in feature_names}])
        proc_input = preprocessor.transform(input_df)
        if hasattr(proc_input, "toarray"):
            proc_input = proc_input.toarray()
        pred = winning_model.predict(proc_input)[0]
        status = "HIGH VIRALITY" if pred == 1 else "LOW ATTRIBUTION"
        color = "#25F4EE" if pred == 1 else "#FE2C55"
        display(HTML(f"""
        <div style="background:#151D30; border:1px solid {color}; padding:12px; border-radius:8px; margin-top:10px; width:260px; text-align:center;">
            <span style="color:#94A3B8; font-size:11px;">ATTRIBUTION OUTCOME</span>
            <h2 style="color:{color}; font-size:20px; margin:4px 0 0 0;">{status}</h2>
        </div>
        """))

predict_btn.on_click(on_click)
display(HTML("<h4 style='color:#25F4EE;'>TikTok Shop Campaign Parameter Controls</h4>"))
display(widgets.GridBox(list(input_widgets.values()), layout=widgets.Layout(grid_template_columns='repeat(2, 300px)', grid_gap='10px')))
display(predict_btn)
display(output_box)
```

## 8. Save Artifacts & Generate Submission CSV

```python
# ==========================================
# 8. Save Artifacts & Generate Submission CSV
# ==========================================
submission_df = pd.DataFrame({
    'Id': np.arange(1, len(winning_preds) + 1),
    'Predicted_Virality': winning_preds
})
submission_path = "/kaggle/working/tiktok_cac_attribution_predictions.csv"
submission_df.to_csv(submission_path, index=False)

model_path = "/kaggle/working/tiktok_virality_model.joblib"
joblib.dump(winning_model, model_path)

print(f"Predictions exported to: {submission_path}")
print(f"Model saved to: {model_path}")
```
